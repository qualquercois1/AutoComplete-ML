# Teorema de Bayes

$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$

## Definição dos Termos

- $\Omega$: Espaço amostral (todas as ocorrências de palavras no dataset).
- $n$: Total de palavras no dataset ($|\Omega| = n$).
- **Evento $A$**: Ocorrência de uma palavra candidata $w$.
- **Evento $B$**: Ocorrência de qualquer palavra que comece com o prefixo $x$.

## Probabilidades

- **Probabilidade a Priori:** $P(A) = \frac{\text{Frequência da palavra } w}{n}$
- **Evidência (Normalização):** $P(B) = \frac{\text{Frequência de palavras com o prefixo } x}{n}$
- **Verossimilhança:** $P(B \mid A) = 1$ se a palavra $w$ começa com o prefixo $x$, ou $0$ caso contrário.
- **Probabilidade a Posteriori:** $P(A \mid B)$ é a probabilidade de a palavra pretendida ser $w$, dado que o prefixo digitado é $x$.

```python
from collections import Counter
import re
```

```python
texto = """A ciencia da computacao é fantastica. A ciencia de dados tambem é incrivel.
Eu gosto de programar em python Eu gosto de programar em c++. 
Inteligencia artificial vai mudar o mundo. Inteligencia artificial aprende com dados."""
```

```python
#calculo de A
palavra_teste = "c"

#regex para separar as palavras
palavras = re.findall(r'\w+', texto.lower())
#dicionario para contar a frequencia das palavras
palavras = dict(Counter(palavras))

print(palavras)
```

```python
#calculo de B
B = [b for b in palavras if b.startswith(palavra_teste)]
B_tam = len(B)

print(B)
print(B_tam)
```

```python
probabilidades = {}
total_palavras = sum(palavras.values())

p_prefixo = B_tam / total_palavras if total_palavras > 0 else 0.0

print(f"N palavras: {total_palavras} | P(B): {p_prefixo:.4f}")

for palavra, frequencia in palavras.items():
    p_palavra = frequencia / total_palavras
    verossimilhanca = 1.0 if palavra.startswith(palavra_teste) else 0.0
    
    if p_prefixo > 0 and verossimilhanca > 0:
        probabilidade_final = (verossimilhanca * p_palavra) / p_prefixo
    else:
        probabilidade_final = 0.0

    probabilidades[palavra] = {
        "p_palavra": p_palavra,
        "verossimilhanca": verossimilhanca,
        "p_prefixo": p_prefixo,
        "probabilidade_final": probabilidade_final
    }
```

```python
probabilidades_ordenadas = sorted(
    probabilidades.items(), 
    key=lambda x: x[1]["probabilidade_final"], 
    reverse=True
)

print("\nProbabilidades finais ordenadas:")
for palavra, info in probabilidades_ordenadas:
    print(f"{palavra}: {info['probabilidade_final']*100:.2f}%")
```

```python
def calcular_probabilidade(prefixo):
    #Texto de exemplo como dataset
    texto = """A ciencia da computacao é fantastica. A ciencia de dados tambem é incrivel.
    Eu gosto de programar em python Eu gosto de programar em c++. 
    Inteligencia artificial vai mudar o mundo. Inteligencia artificial aprende com dados."""

    #calculo de A
    #regex para separar as palavras
    palavras = re.findall(r'\w+', texto.lower())
    #dicionario para contar a frequencia das palavras
    palavras = dict(Counter(palavras))

    #calculo de B
    B = [b for b in palavras if b.startswith(prefixo)]
    B_tam = len(B)

    probabilidades = {}
    #calculo de n
    total_palavras = sum(palavras.values())

    #calculo de P(B)
    p_prefixo = B_tam / total_palavras if total_palavras > 0 else 0.0

    for palavra, frequencia in palavras.items():
        #calculo de P(A)
        p_palavra = frequencia / total_palavras
        #calculo de P(B|A)
        verossimilhanca = 1.0 if palavra.startswith(prefixo) else 0.0
        
        if p_prefixo > 0 and verossimilhanca > 0:
            #calculo de P(A|B)
            probabilidade_final = (verossimilhanca * p_palavra) / p_prefixo
        else:
            probabilidade_final = 0.0

        probabilidades[palavra] = {
            "p_palavra": p_palavra,
            "verossimilhanca": verossimilhanca,
            "p_prefixo": p_prefixo,
            "probabilidade_final": probabilidade_final
        }

    probabilidades_ordenadas = sorted(
        probabilidades.items(), 
        key=lambda x: x[1]["probabilidade_final"], 
        reverse=True
    )
    
    return probabilidades_ordenadas

```
